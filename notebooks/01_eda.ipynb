```python
   from forest_fire_risk.data import load_track_a

   df = load_track_a()

   # one row per sample: its label and the last day of its 30-day window
   s = df.groupby("sample_id").agg(label=("label", "first"), end_date=("time", "max")).reset_index()
   s["year"] = s["end_date"].dt.year

   # how many non-fire (0) and fire (1) samples per year
   print(s.groupby(["year", "label"]).size().unstack(fill_value=0))
```
